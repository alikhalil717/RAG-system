```python
import numpy as np
import joblib
from sentence_transformers import SentenceTransformer
from sklearn.datasets import fetch_20newsgroups
import pandas as pd

# تعريف الدوال أولاً
def clean_text(text):
    """تنظيف النص"""
    return text.strip()

def cosine_similarity(v1, array_of_vectors):
    """حساب التشابه جيب التمام"""
    # تحويل v1
    if hasattr(v1, "detach"):
        v1 = v1.detach().cpu().numpy()
    v1 = np.asarray(v1, dtype=np.float32).ravel()

    # تحويل array_of_vectors
    if hasattr(array_of_vectors, "detach"):
        array_of_vectors = array_of_vectors.detach().cpu().numpy()
    A = np.asarray(array_of_vectors, dtype=np.float32)
    
    # حالة متجه واحد
    if A.ndim == 1:
        A = A.ravel()  # تصحيح: ravel وليس reval
        denom = np.linalg.norm(v1) * np.linalg.norm(A)
        return float(0.0 if denom == 0 else (v1 @ A) / denom)
    
    # حالة مصفوفة متجهات
    A = np.atleast_2d(A)
    v1_norm = np.linalg.norm(v1)
    A_norms = np.linalg.norm(A, axis=1)
    denom = v1_norm * A_norms
    
    with np.errstate(divide='ignore', invalid='ignore'):
        sims = (A @ v1) / np.where(denom == 0, 1.0, denom)
    sims[denom == 0] = 0.0
    return sims.tolist()

def top_k_indices(scores, k):
    """إرجاع indices لأعلى k قيمة"""
    return np.argsort(scores)[::-1][:k].tolist()

def retrieve_relevant(que, embeddings, model, df, data_fetch, top_k=5):
    """استرجاع المستندات الأكثر صلة"""
    # تنظيف الاستعلام وتشفيره
    que = clean_text(que)
    que_em = model.encode([que])  # تمرير كقائمة
    
    # حساب التشابه
    similarities = cosine_similarity(que_em, embeddings)
    
    # الحصول على top-k indices
    top_results = top_k_indices(similarities, top_k)
    
    # طباعة النتائج
    print(f"Query: {que}")
    print("=" * 60)
    for idx in top_results:
        print(f"\nDocument Index: {idx}")
        print(f"Document: {df.iloc[idx]['text'][:300]}...")
        print(f"Category: {data_fetch.target_names[df.iloc[idx]['category']]}")
        print(f"Similarity Score: {similarities[idx]:.4f}")
        print("-" * 40)
    
    return top_results

# ============================================
# الآن تشغيل الكود
# ============================================

# 1. تحميل البيانات
print("جاري تحميل البيانات...")
data_fetch = fetch_20newsgroups(
    subset='train',
    shuffle=True,
    random_state=42,
    data_home='./dataset'
)

# 2. إنشاء DataFrame
df = pd.DataFrame({
    'text': data_fetch.data,
    'category': data_fetch.target
})

# 3. تحميل النموذج
print("جاري تحميل النموذج...")
model_name = "BAAI/bge-base-en-v1.5"
model = SentenceTransformer(model_name, device='cpu')

# 4. تحميل embeddings
print("جاري تحميل embeddings...")
embeddings = joblib.load('embeddings.joblib')
print(f"Shape of embeddings: {embeddings.shape}")

# 5. تنفيذ البحث
print("\n" + "=" * 60)
example_query = "space exploration"
results = retrieve_relevant(example_query, embeddings, model, df, data_fetch, top_k=2)
```
